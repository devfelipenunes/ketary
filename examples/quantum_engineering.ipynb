# Ketary: first quantum engineering workflow

Run after installing Ketary locally with Python 3.12. This notebook uses ideal logical circuits; no hardware or implicit execution.

```python
import ketary as q
c = q.bell()
r = c.run()
print(c.visualize.text())
print(r.math.dirac())
```

```python
contract = q.Contract(2, probabilities={"00": 0.5, "11": 0.5}, expectations={"XX": 1, "ZZ": 1})
print(contract.verify(r).require().render())
```

```python
print(r.math.density_matrix([0]))
print(r.math.purity([0]))
print(r.visualize.bloch(0))
```

```python
from pathlib import Path
chart = r.visualize.histogram(k=2)
print(chart[:100])
# To save a chart explicitly: Path('bell.svg').write_text(chart, encoding='utf-8')
```

```python
observed = c.copy().z(0)
print(contract.verify(observed.run()).render())
assert q.grover('101').run().math.probability('101') > 0.9
```
